# Class 4 - Queues
A queue is an ordered collection of elements where items are added and removed from *opposite* ends, commonly referered to the as the "front" and "back" or 
"rear" of the queue. The ordering principle is referrred as the FIFO (FIrst In, First Out) or "First come, first served".

```python
# Simmplified Queue (relying on built-ins) 
class Queue:
    def __init__(self):
        self.items = []

    def enqueue(self, element):
        self.items.insert(0, element)

    def dequeue(self):
        return self.items.pop()

    def peek(self):
        return self.items[len(self.items)-1]

    def size(self):
        return len(self.items)

    def is_empty(self):
        return self.items == []

    def __str__(self):
        return f"Queue - {self.items}"
```

```python
q1 = Queue()
q1.enqueue("Edwin")
q1.enqueue("Chante")
q1.enqueue("Jey")
q1.enqueue("Austin")
q1.enqueue("Luis")
print(q1)
q1.dequeue()
print(q1)
q1.dequeue()
print(q1)
q1.dequeue()
print(q1)
q1.dequeue()
print(q1)
hold = q1.dequeue()
print(q1)
print(hold)
```
