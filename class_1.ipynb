# OOP (Object Oriented Programming)
Python is an object-oriented programming, allowing us to structure code using classes and objects for a better organizations and reusability.

## Advantages
-Provides clear structure to programs
-Makes code easier to maintain, reuse and debug
-Helps keep your doe DRY(Do not Repeat Yourself)
-Allows to rebuild reusable applications with less code 

## CLasses and Objects
A class defines what an object should look like and it's behavior
An object is created based on that class.
For example:

    Class -> Fruit
    Objects -> Apple, Banana, Mango
    Class -> Car
    Objects -> Audi, Tesla, Ford

#### the four pillars of OOP
1. Encapsulation
2. Abstraction
3. Inheritance
4. Polymorphism

Example of a class: 

    class PostListView(ListView):
    

```python
class Person:
    #Constructor
    def __init__(self, name, last_name, age):
        self.name = name
        self.last_name = last_name
        self.age = age
        self.is_active = False

    def greeting(self):
        return f"Hello from {self.name}"

    def __str__(self):
        return f"Person (name = {self.name}, last_name = {self.last_name}, age = {self.age}, is_active = {self.is_active})"
```

```python
p1 = Person("Luis", "Renteria", 25)
p1.is_active = True
print(p1)
p1.greeting()
```

# Assignment 1 : Find Even Numbers

Find the total number of numbers from a collection that divide by 2 the modulus is 0

Examples:
1. Given [3,4,10,15,999,2] then the output should be; total = 3
2. Given [2,4,6,10,3,7] then the output should be: total = 4

Criteria:
1. Your function should receive a list and return an integer

Hint: 

    - '%' is hte math operand to get the modulus from a division 
    -2 % 2 = 0

The collection is: 
numbers = [3,5,27,20,6,3,5,60]


```python
def count_even_numbers(numbers):            #Create a function that receives a list called numbers
    total = 0                               #Start the count at zero.

    for number in numbers:                  #Look at each number in the list, one at a time
        if number % 2 == 0:                 #Check whether dividing that number by 2 leaves a remainder of zero. If it does, the number is even
            total += 1                      #Add one to the count. This means the same thing as total = total + 1

    return total                            #Send the final count back as an integer


numbers = [3, 5, 27, 20, 6, 3, 5, 60]

result = count_even_numbers(numbers)
print("total =", result)                    #Print the text "total =", followed by the value stored in result
```

# Assignment 1: largest num

Create a function that will help us to determine that largest number from a list of numbers

Criteria
1. The function should receive a list and return the largest number
2. You can test with a list of any values from your selection 

```python
def find_largest_number(numbers):                   #Create a function that receives a list of numbers
    largest = numbers[0]                            #Start by treating the first number as the largest. Python starts counting positions at 0, so here that number is 3

    for number in numbers:                          #Check each number in the list
        if number > largest:                        #Ask: “Is this number bigger than the largest one found so far?”
            largest = number                        #If yes, remember this number as the new largest.

    return largest                                  #Send the largest number back


numbers = [3, 5, 27, 20, 6, 3, 5, 60]

result = find_largest_number(numbers)               #Give the list called numbers to the function find_largest_number. Store the answer it returns in a variable called result
print("Largest number =", result)                   #Display the words Largest number =, followed by the value stored in result
```

```python
def find_largest_number(numbers):               
    largest = numbers[0]

    for number in numbers:                          #Go through the list one item at a time. Each current item is temporarily called number
        if number > largest:                        #Check whether the current item is greater than the item stored in largest. Because these items are strings, Python compares them as text
            largest = number                        #If the condition is true, replace largest with the current item.

    return largest                                  #Send the value of largest back to whoever called this function. This also ends the function


numbers = ["1", "2"]                                # The twist: Create a list called numbers containing two strings. The quotation marks make "1" and "2" text rather than integers

result = find_largest_number(numbers)               #Run the function using the list called numbers. Store the value it returns in a variable called result

print("Largest number =", result)                   #Display the words "Largest number =", followed by the value stored in result.
print("Type of result =", type(result))             #Find the data type of result using type(result). Then display "Type of result =" followed by that data type
```
