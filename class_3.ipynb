## Stacks
A stack is an ordered collection of elements where items are added and removed from one end, commonly referred to as the "top". The ordering principle is referred as LIFO (last in, First out)

```python
# Simplified Version (relying on built-ins functions)

class Stack:
    def __init__(self):
        self.items = []

    def push(self, elements):
        self.items.append(elements)

    def pop(self):
        return self.items.pop()

    def peek(self):
        return self.items[len(self.items) - 1]

    def is_empty(self):
        return self.items == []

    def __str__(self):
        return f"Stack - {self.items}"
```

```python
s1 = Stack()
s1.push("A")
s1.push("B")
s1.push("C")
print(s1.peek())
print(s1.is_empty())
print(s1)
hold = s1.pop()
print(s1)
print(hold)
```

```python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, elements):
        self.items.append(elements)

    def pop(self):
        return self.items.pop()

    def peek(self):
        return self.items[len(self.items) - 1]

    def is_empty(self):
        return self.items == []

    def __str__(self):
        return f"Stack - {self.items}"


def invert_string(word):
    inverted_word  = ""  

    stack = Stack()           
    for letter in word:
        stack.push(letter)

    while not stack.is_empty():
        inverted_word += stack.pop()   

    return inverted_word

print(invert_string("Rafael"))
print(invert_string("stars"))
print(invert_string("cool"))
```
