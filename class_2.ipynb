# Mini Challenge 2: Sum Numbers greater than 5
Find the sum of all numbers greater than 5

*Examples:*
1. Given [2,7,8,3] Then the output should be total = 15
2. Given [1,2,3] Then the output should be total = 0

*Criteria*
1. Your functions should receive a list and return an integer.

The collection is: [3,5,22,2,1,6,1,6,60]


```python
numbers = [3, 5, 22, 2, 1.6, 1, 6, 60]

def sum_numbers(numbers):
    total = 0
    for number in numbers: 
        if number > 5:
            total += number
    return total

print(sum_numbers(numbers))
```

## Mini Challenge 3 Anagram Check

Create a function that takes two strings as parameters and returns True if they are anagrams of each other. False otherwise

    Example:
    1. heart and earth
    2. scar and cars
    3. start and rats
    etc

*Criteria:*
Two stringss are anagrams of each other if and only if all of the criteria below are met:

    1. The must use all the same characters
    2. They must be of exactly the same length (in chars)
    3. AI and Sorted functions is not allowed


```python
def letters_in_word(word1, word2):
    if len(word1) != len(word2):
        return False

    for letter in word1:
        if word1.count(letter) != word2.count(letter):
            return False
    return True

print(letters_in_word("heart", "earth"))
print(letters_in_word("scar", "car"))
print(letters_in_word("star", "rats"))
```

```python
def are_anagrams(str1, str2):
    counter = 0
    if len(str1) == len(str2):
        for letter1 in str1:
            for letter2 in str2:
                if letter1 == letter2:
                    counter += 1
        if len(str1) != counter:
            return False
        else:
            return True
    else:
        return False
are_anagrams("rats", "star")
```
